```python
import os
import cv2
import matplotlib.pyplot as plt

IMG_DIR = "/teamspace/studios/this_studio/workspace/Datasets/Dota/images"
LABEL_DIR = "/teamspace/studios/this_studio/workspace/Datasets/Dota/labelTxt"

def parse_label(label_path):
    boxes = []
    
    with open(label_path, "r") as f:
        lines = f.readlines()
    
    for line in lines:
        parts = line.strip().split()
        
        # 🚨 Skip metadata lines
        if len(parts) < 9:
            continue
        
        # Skip lines that don't start with numbers
        try:
            coords = list(map(float, parts[:8]))
        except:
            continue
        
        cls = parts[8]
        
        xs = coords[0::2]
        ys = coords[1::2]
        
        x_min = int(min(xs))
        y_min = int(min(ys))
        x_max = int(max(xs))
        y_max = int(max(ys))
        
        boxes.append((x_min, y_min, x_max, y_max, cls))
    
    return boxes


def visualize_sample():
    img_name = os.listdir(IMG_DIR)[0]  # pick first image
    
    img_path = os.path.join(IMG_DIR, img_name)
    label_path = os.path.join(LABEL_DIR, img_name.replace(".png", ".txt"))
    
    image = cv2.imread(img_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    boxes = parse_label(label_path)
    
    for (x1, y1, x2, y2, cls) in boxes:
        cv2.rectangle(image, (x1, y1), (x2, y2), (255, 0, 0), 2)
        cv2.putText(image, cls, (x1, y1-5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,0,0), 1)
    
    plt.figure(figsize=(10,10))
    plt.imshow(image)
    plt.axis("off")
    plt.title(img_name)
    plt.show()


visualize_sample()
```

```python
import torch
import clip
from PIL import Image

device = "cuda" if torch.cuda.is_available() else "cpu"

model, preprocess = clip.load("ViT-B/32", device=device)

print("CLIP loaded on:", device)
```

```python
import os
import cv2

IMG_DIR = "/teamspace/studios/this_studio/workspace/Datasets/Dota/images"
LABEL_DIR = "/teamspace/studios/this_studio/workspace/Datasets/Dota/labelTxt"

def get_dota_crops(max_images=5):
    crops = []
    
    img_files = os.listdir(IMG_DIR)[:max_images]
    
    for img_name in img_files:
        img_path = os.path.join(IMG_DIR, img_name)
        label_path = os.path.join(LABEL_DIR, img_name.replace(".png", ".txt"))
        
        image = cv2.imread(img_path)
        
        if not os.path.exists(label_path):
            continue
        
        with open(label_path, "r") as f:
            lines = f.readlines()
        
        for line in lines:
            parts = line.strip().split()
            
            if len(parts) < 9:
                continue
            
            try:
                coords = list(map(float, parts[:8]))
            except:
                continue
            
            cls = parts[8]
            
            xs = coords[0::2]
            ys = coords[1::2]
            
            x1, y1 = int(min(xs)), int(min(ys))
            x2, y2 = int(max(xs)), int(max(ys))
            
            crop = image[y1:y2, x1:x2]
            
            if crop.size == 0:
                continue
            
            crops.append((crop, cls))
    
    return crops
```

```python
VIS_BASE = "/teamspace/studios/this_studio/workspace/Datasets/Visdrone/VisDrone2019-DET-train"
IMG_DIR_V = VIS_BASE + "/images"
ANN_DIR_V = VIS_BASE + "/annotations"

VISDRONE_CLASSES = {
    1: "pedestrian",
    2: "person",
    3: "bicycle",
    4: "car",
    5: "van",
    6: "truck",
    7: "tricycle",
    8: "awning-tricycle",
    9: "bus",
    10: "motor"
}

def get_visdrone_crops(max_images=5):
    crops = []
    
    img_files = os.listdir(IMG_DIR_V)[:max_images]
    
    for img_name in img_files:
        img_path = os.path.join(IMG_DIR_V, img_name)
        ann_path = os.path.join(ANN_DIR_V, img_name.replace(".jpg", ".txt"))
        
        image = cv2.imread(img_path)
        
        with open(ann_path, "r") as f:
            lines = f.readlines()
        
        for line in lines:
            parts = line.strip().split(",")
            
            x, y, w, h = map(int, parts[:4])
            cls_id = int(parts[5])
            
            if cls_id == 0:
                continue
            
            cls = VISDRONE_CLASSES.get(cls_id, "unknown")
            
            crop = image[y:y+h, x:x+w]
            
            if crop.size == 0:
                continue
            
            crops.append((crop, cls))
    
    return crops
```

```python
import numpy as np

def encode_crops(crops):
    image_features = []
    labels = []
    
    for crop, label in crops:
        pil_img = Image.fromarray(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
        
        image_input = preprocess(pil_img).unsqueeze(0).to(device)
        
        with torch.no_grad():
            feat = model.encode_image(image_input)
            feat = feat / feat.norm(dim=-1, keepdim=True)
        
        image_features.append(feat.cpu().numpy())
        labels.append(label)
    
    return np.vstack(image_features), labels
```

```python
dota_crops = get_dota_crops()
vis_crops = get_visdrone_crops()

dota_feats, dota_labels = encode_crops(dota_crops)
vis_feats, vis_labels = encode_crops(vis_crops)

print("DOTA features:", dota_feats.shape)
print("VisDrone features:", vis_feats.shape)
```

```python
import torch
from PIL import Image
from transformers import CLIPProcessor, CLIPModel
import os
from tqdm import tqdm
import numpy as np
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device)
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
```

```python
image_folder = "/workspace/datasets/dota/images"
class_names = [
    "plane",
    "ship",
    "vehicle",
    "building",
    "harbor"
]
```

```python
text_inputs = processor(
    text=class_names,
    return_tensors="pt",
    padding=True
).to(device)

with torch.no_grad():
    text_features = model.get_text_features(**text_inputs)
    text_features = text_features / text_features.norm(dim=-1, keepdim=True)
```

```python
image_features_list = []
image_names = []

image_files = os.listdir(image_folder)

for img_name in tqdm(image_files[:300]):   # increase from 50 → 300  
    img_path = os.path.join(image_folder, img_name)
    
    try:
        image = Image.open(img_path).convert("RGB")
        
        inputs = processor(images=image, return_tensors="pt").to(device)

        with torch.no_grad():
            image_features = model.get_image_features(**inputs)
            image_features = image_features / image_features.norm(dim=-1, keepdim=True)

        image_features_list.append(image_features.cpu().numpy())
        image_names.append(img_name)

    except:
        continue

image_features_array = np.vstack(image_features_list)
```

```python
similarity = image_features_array @ text_features.cpu().numpy().T
```

```python
predictions = similarity.argmax(axis=1)

for i in range(10):
    print(f"{image_names[i]} → {class_names[predictions[i]]}")
```

```python
#Computing Drift
```

```python
class_embeddings = {}

for i in range(len(combined_features)):
    cls = combined_labels[i]
    domain = domain_labels[i]
    
    if cls not in class_embeddings:
        class_embeddings[cls] = {"satellite": [], "uav": []}
    
    class_embeddings[cls][domain].append(combined_features[i])
```

```python
import numpy as np

class_centroids = {}

for cls in class_embeddings:
    sat = np.array(class_embeddings[cls]["satellite"])
    uav = np.array(class_embeddings[cls]["uav"])
    
    if len(sat) == 0 or len(uav) == 0:
        continue
    
    sat_centroid = sat.mean(axis=0)
    uav_centroid = uav.mean(axis=0)
    
    class_centroids[cls] = (sat_centroid, uav_centroid)
```

```python
from numpy.linalg import norm

drift_scores = {}

for cls, (sat_c, uav_c) in class_centroids.items():
    distance = norm(sat_c - uav_c)
    drift_scores[cls] = distance
```

```python
for cls, score in drift_scores.items():
    print(f"{cls}: Drift Score = {score:.4f}")
```

```python
from numpy.linalg import norm

classes = list(class_centroids.keys())
n = len(classes)

distance_matrix = np.zeros((n, n))

for i, cls_sat in enumerate(classes):
    for j, cls_uav in enumerate(classes):
        
        sat_vec = class_centroids[cls_sat][0]   # satellite centroid
        uav_vec = class_centroids[cls_uav][1]   # uav centroid
        
        distance_matrix[i, j] = norm(sat_vec - uav_vec)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

im = plt.imshow(distance_matrix)

plt.colorbar(im, label="Distance")

plt.xticks(range(n), classes, rotation=45)
plt.yticks(range(n), classes)

plt.xlabel("UAV Classes")
plt.ylabel("Satellite Classes")

plt.title("Cross-Domain Distance Heatmap")

plt.tight_layout()
plt.show()
```
